```python
import cv2
from ultralytics import YOLO

# Load the lightweight Nano model
model = YOLO("yolo11n.pt")

# Initialize webcam (0 is usually the default built-in camera)
cap = cv2.VideoCapture(0)

while cap.isOpened():
    success, frame = cap.read()
    if not success:
        break

    # Run YOLO11 Nano inference on the current frame
    # stream=True utilizes a generator to efficiently save memory
    results = model(frame, stream=True)

    for r in results:
        # Get the frame annotated with bounding boxes and labels
        annotated_frame = r.plot()

    # Display the real-time AI vision stream
    cv2.imshow("YOLO Nano Real-Time Detection", annotated_frame)

    # Break the loop if the 'q' key is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
